```python
from pyspark.sql.types import *

schema = StructType([
    StructField("id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True),
    StructField("salary", DoubleType(), True)
])

```

```python
df = spark.read 
    .option("header", "true") 
    .schema(schema) 
    .csv("/path/to/employees.csv")

df.write 
    .mode("overwrite") 
    .saveAsTable("employees")
```

```python
spark.sql("""
CREATE OR REPLACE TEMP VIEW orders AS
SELECT * FROM VALUES
    (101, 1, '2024-01-15', 250.00),
    (102, 1, '2024-02-20', 150.00),
    (103, 2, '2024-01-10', 300.00),
    (104, 2, '2024-03-05', 450.00),
    (105, 3, '2024-02-01', 100.00)
AS t(order_id, customer_id, order_date, amount)
""")

spark.sql("""
WITH ranked AS (
    SELECT *,
           ROW_NUMBER() OVER (
               PARTITION BY customer_id
               ORDER BY CAST(order_date AS DATE) DESC, order_id DESC
           ) AS rn
    FROM orders
)
SELECT order_id, customer_id, order_date, amount
FROM ranked
WHERE rn = 1
ORDER BY customer_id
""").show()
```
